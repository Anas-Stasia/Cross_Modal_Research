```python
import pandas as pd
import numpy as np
from pathlib import Path

# ============================================================
# PROJECT PATHS
# ============================================================

# Notebook is inside:
# CROSS_MODAL_RESEARCH/src/data_cleaning.ipynb

RAW_DIR = Path("../data/raw")
PROCESSED_DIR = Path("../data/processed")



# Raw files
METADATA_PATH = RAW_DIR / "metadata" / "meta_data.csv"
TEXT_PATH = RAW_DIR / "text" / "text_embeddings.csv"
AUDIO_PATH = RAW_DIR / "audio" / "audio_embeddings.csv"
VIDEO_PATH = RAW_DIR / "video" / "video_embeddings.csv"

print("Metadata:", METADATA_PATH)
print("Text:", TEXT_PATH)
print("Audio:", AUDIO_PATH)
print("Video:", VIDEO_PATH)
```

```python
metadata = pd.read_csv(METADATA_PATH)
text = pd.read_csv(TEXT_PATH)
audio = pd.read_csv(AUDIO_PATH)
video = pd.read_csv(VIDEO_PATH)

print("Metadata:", metadata.shape)
print("Text:", text.shape)
print("Audio:", audio.shape)
print("Video:", video.shape)
```

```python
print("METADATA COLUMNS")
print(metadata.columns)

print("\nTEXT COLUMNS")
print(text.columns)

print("\nAUDIO COLUMNS")
print(audio.columns)

print("\nVIDEO COLUMNS")
print(video.columns)
```

```python
print("Unique metadata IDs:", metadata["id_hash"].nunique())
print("Unique text IDs:", text["id_hash"].nunique())
print("Unique audio IDs:", audio["id_hash"].nunique())
print("Unique video IDs:", video["id_hash"].nunique())

print("\nID uniqueness:")
print("Metadata:", metadata["id_hash"].is_unique)
print("Text:", text["id_hash"].is_unique)
print("Audio:", audio["id_hash"].is_unique)
print("Video:", video["id_hash"].is_unique)
```

```python
print(
    metadata[
        ["plays", "likes", "comments", "shares"]
    ].describe()
)
```

```python
metadata_ids = set(metadata["id_hash"])
text_ids = set(text["id_hash"])
audio_ids = set(audio["id_hash"])
video_ids = set(video["id_hash"])

all_three = text_ids & audio_ids & video_ids

all_four = (
    metadata_ids
    & text_ids
    & audio_ids
    & video_ids
)

print("Metadata:", len(metadata_ids))
print("Text:", len(text_ids))
print("Audio:", len(audio_ids))
print("Video:", len(video_ids))

print("\nPAIRWISE / MULTIMODAL INTERSECTIONS")

print(
    "Text + Audio:",
    len(text_ids & audio_ids)
)

print(
    "Text + Video:",
    len(text_ids & video_ids)
)

print(
    "Audio + Video:",
    len(audio_ids & video_ids)
)

print(
    "Text + Audio + Video:",
    len(all_three)
)

print(
    "Metadata + Text + Audio + Video:",
    len(all_four)
)
```

```python
duplicate_metadata = metadata[
    metadata["id_hash"].duplicated(keep=False)
].sort_values("id_hash")

print("Duplicate metadata rows:", duplicate_metadata.shape)

print(
    duplicate_metadata.head(20).to_string(index=False)
)
```

```python
print(
    metadata["id_hash"]
    .value_counts()
    .value_counts()
    .sort_index()
)
```

```python
metadata["has_all_embeddings"] = (
    metadata["id_hash"].isin(all_four)
)

print(
    metadata.groupby("has_all_embeddings")["plays"]
    .describe()
)
```

```python
subset = metadata.drop_duplicates("id_hash").copy()

subset["has_video_embedding"] = (
    subset["id_hash"].isin(video_ids)
)

print(
    subset.groupby("has_video_embedding")[
        [
            "plays",
            "likes",
            "comments",
            "shares",
            "author_followers"
        ]
    ].median()
)
```

```python
print(
    pd.crosstab(
        subset["has_video_embedding"],
        subset["category"],
        normalize="index"
    )
)
```

```python
common_ids = list(all_four)

text_common = text[
    text["id_hash"].isin(common_ids)
].copy()

audio_common = audio[
    audio["id_hash"].isin(common_ids)
].copy()

video_common = video[
    video["id_hash"].isin(common_ids)
].copy()

print("Text:", text_common.shape)
print("Audio:", audio_common.shape)
print("Video:", video_common.shape)
```

```python
text_values = text_common.iloc[:, 1:].values.astype(np.float32)
audio_values = audio_common.iloc[:, 1:].values.astype(np.float32)
video_values = video_common.iloc[:, 1:].values.astype(np.float32)

print("Text NaNs:", np.isnan(text_values).sum())
print("Audio NaNs:", np.isnan(audio_values).sum())
print("Video NaNs:", np.isnan(video_values).sum())

print("\nText infinities:", np.isinf(text_values).sum())
print("Audio infinities:", np.isinf(audio_values).sum())
print("Video infinities:", np.isinf(video_values).sum())

print("\nText finite:", np.isfinite(text_values).all())
print("Audio finite:", np.isfinite(audio_values).all())
print("Video finite:", np.isfinite(video_values).all())
```

```python
video_invalid_mask = (
    ~np.isfinite(video_values)
).any(axis=1)

bad_video_ids = video_common.loc[
    video_invalid_mask,
    "id_hash"
].tolist()

print("Number of problematic videos:", len(bad_video_ids))

print("\nProblematic video IDs:")

for video_id in bad_video_ids:
    print(video_id)
```

```python
video_nan_counts = (
    video_common.iloc[:, 1:]
    .isna()
    .sum(axis=1)
)

print("video_common shape:", video_common.shape)

print(
    "Unique video IDs:",
    video_common["id_hash"].nunique()
)

print(
    "Duplicate IDs:",
    video_common["id_hash"].duplicated().sum()
)

print("\nTotal NaNs:")
print(video_common.iloc[:, 1:].isna().sum().sum())

print("\nNaNs per video:")
print(video_nan_counts.describe())

print("\nMissingness groups:")

print(
    "0:",
    (video_nan_counts == 0).sum()
)

print(
    "1-10:",
    (
        (video_nan_counts >= 1)
        & (video_nan_counts <= 10)
    ).sum()
)

print(
    "11-100:",
    (
        (video_nan_counts >= 11)
        & (video_nan_counts <= 100)
    ).sum()
)

print(
    "101-500:",
    (
        (video_nan_counts >= 101)
        & (video_nan_counts <= 500)
    ).sum()
)

print(
    ">500:",
    (video_nan_counts > 500).sum()
)
```

```python
video_nan_counts = (
    video_common.iloc[:, 1:]
    .isna()
    .sum(axis=1)
)

print("video_common shape:", video_common.shape)

print(
    "Unique video IDs:",
    video_common["id_hash"].nunique()
)

print(
    "Duplicate IDs:",
    video_common["id_hash"].duplicated().sum()
)

print("\nTotal NaNs:")
print(video_common.iloc[:, 1:].isna().sum().sum())

print("\nNaNs per video:")
print(video_nan_counts.describe())

print("\nMissingness groups:")

print(
    "0:",
    (video_nan_counts == 0).sum()
)

print(
    "1-10:",
    (
        (video_nan_counts >= 1)
        & (video_nan_counts <= 10)
    ).sum()
)

print(
    "11-100:",
    (
        (video_nan_counts >= 11)
        & (video_nan_counts <= 100)
    ).sum()
)

print(
    "101-500:",
    (
        (video_nan_counts >= 101)
        & (video_nan_counts <= 500)
    ).sum()
)

print(
    ">500:",
    (video_nan_counts > 500).sum()
)
```

```python
metadata_clean["scrape_date"] = pd.to_datetime(
    metadata_clean["date_scraping"],
    format="%d_%m_%Y",
    errors="coerce"
)
```

```python
metadata_clean = metadata_clean.sort_values(
    ["id_hash", "scrape_date"]
)
```

```python
metadata_final = (
    metadata_clean
    .drop_duplicates(
        "id_hash",
        keep="last"
    )
    .copy()
)
```
